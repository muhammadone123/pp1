```python
# =================================
# CHECK OUTLIERS USING IQR METHOD
# =================================

# 1. Select Numerical Columns
num_cols = df.select_dtypes(include=np.number).columns

# 2. Check Outliers in Each Column
for col in num_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]

    print(f"\nColumn: {col}")
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Total Outliers:", len(outliers))

# 3. Visualize Outliers Using Boxplot
plt.figure(figsize=(12, 6))
sns.boxplot(data=df[num_cols])
plt.xticks(rotation=45)
plt.title("Outlier Detection Using Boxplot")
plt.show()

# 4. Display Dataset
display(df)
```

```python
import numpy as np
import pandas as pd
import plotly.express as px
import streamlit as st
 
st.set_page_config(page_title="Sales Dashboard", page_icon="📊", layout="wide")
 
# ---------------------------------------------------------------
# 1. DATA  (sample data; replace with your own file, see note below)
# ---------------------------------------------------------------
@st.cache_data
def load_data() -> pd.DataFrame:
    # To use your own file instead, uncomment these lines:
    # df = pd.read_csv("your_file.csv", parse_dates=["Date"])
    # return df
    rng = np.random.default_rng(42)
    n = 3000
    dates = pd.to_datetime(
        rng.integers(
            pd.Timestamp("2024-01-01").value // 10**9,
            pd.Timestamp("2025-12-31").value // 10**9,
            n,
        ),
        unit="s",
    ).normalize()
 
    category_map = {
        "Electronics": ["Phones", "Laptops", "Accessories"],
        "Fashion": ["Men", "Women", "Kids"],
        "Home": ["Furniture", "Kitchen", "Decor"],
    }
    categories = rng.choice(list(category_map), n)
    sub_categories = [rng.choice(category_map[c]) for c in categories]
 
    df = pd.DataFrame(
        {
            "Date": dates,
            "Order ID": np.arange(1, n + 1),
            "Customer ID": rng.integers(1, 900, n),
            "Region": rng.choice(["North", "South", "East", "West"], n),
            "Segment": rng.choice(["Consumer", "Corporate", "Home Office"], n),
            "Category": categories,
            "Sub-Category": sub_categories,
            "Sales": rng.gamma(2.5, 120, n).round(2),
            "Discount": rng.choice([0, 0.05, 0.1, 0.2, 0.3], n),
        }
    )
    df["Profit"] = (df["Sales"] * rng.normal(0.18, 0.12, n) - df["Sales"] * df["Discount"] * 0.3).round(2)
    return df.sort_values("Date").reset_index(drop=True)
 
 
df = load_data()
 
# ---------------------------------------------------------------
# 2. SLICERS (sidebar filters) + TIMELINE (date range slider)
# ---------------------------------------------------------------
st.sidebar.title("🎛️ Slicers")
 
min_d, max_d = df["Date"].min().date(), df["Date"].max().date()
start_date, end_date = st.sidebar.slider(
    "📅 Timeline",
    min_value=min_d,
    max_value=max_d,
    value=(min_d, max_d),
    format="DD MMM YYYY",
)
 
regions = st.sidebar.multiselect("Region", sorted(df["Region"].unique()), default=sorted(df["Region"].unique()))
categories = st.sidebar.multiselect("Category", sorted(df["Category"].unique()), default=sorted(df["Category"].unique()))
segments = st.sidebar.multiselect("Segment", sorted(df["Segment"].unique()), default=sorted(df["Segment"].unique()))
 
mask = (
    df["Date"].between(pd.Timestamp(start_date), pd.Timestamp(end_date))
    & df["Region"].isin(regions)
    & df["Category"].isin(categories)
    & df["Segment"].isin(segments)
)
fdf = df[mask]
 
st.title("📊 Sales Dashboard")
st.caption(f"Showing {len(fdf):,} orders from {start_date:%d %b %Y} to {end_date:%d %b %Y}")
 
if fdf.empty:
    st.warning("No data for the selected filters. Please change the slicers.")
    st.stop()
 
# ---------------------------------------------------------------
# 3. SIX KPI CARDS
# ---------------------------------------------------------------
total_sales = fdf["Sales"].sum()
total_profit = fdf["Profit"].sum()
total_orders = fdf["Order ID"].nunique()
total_customers = fdf["Customer ID"].nunique()
aov = total_sales / total_orders
margin = total_profit / total_sales * 100
 
k1, k2, k3, k4, k5, k6 = st.columns(6)
k1.metric("💰 Total Sales", f"${total_sales:,.0f}")
k2.metric("📈 Total Profit", f"${total_profit:,.0f}")
k3.metric("🧾 Orders", f"{total_orders:,}")
k4.metric("👥 Customers", f"{total_customers:,}")
k5.metric("🛒 Avg Order Value", f"${aov:,.2f}")
k6.metric("📊 Profit Margin", f"{margin:.1f}%")
 
st.divider()
 
# ---------------------------------------------------------------
# 4. SIX CHARTS
# ---------------------------------------------------------------
# Chart 1: Monthly sales & profit trend (line)
monthly = fdf.groupby(pd.Grouper(key="Date", freq="MS"))[["Sales", "Profit"]].sum().reset_index()
fig1 = px.line(monthly, x="Date", y=["Sales", "Profit"], markers=True, title="Monthly Sales & Profit Trend")
fig1.update_layout(legend_title_text="", yaxis_title="Amount ($)")
 
# Chart 2: Sales by category (bar)
cat = fdf.groupby("Category", as_index=False)["Sales"].sum().sort_values("Sales", ascending=False)
fig2 = px.bar(cat, x="Category", y="Sales", color="Category", text_auto=".2s", title="Sales by Category")
fig2.update_layout(showlegend=False)
 
# Chart 3: Sales by region (donut)
reg = fdf.groupby("Region", as_index=False)["Sales"].sum()
fig3 = px.pie(reg, names="Region", values="Sales", hole=0.5, title="Sales Share by Region")
 
# Chart 4: Profit by sub-category (horizontal bar)
sub = fdf.groupby("Sub-Category", as_index=False)["Profit"].sum().sort_values("Profit")
fig4 = px.bar(sub, x="Profit", y="Sub-Category", orientation="h", color="Profit",
              color_continuous_scale="RdYlGn", title="Profit by Sub-Category")
 
# Chart 5: Discount vs profit (scatter)
sample = fdf.sample(min(len(fdf), 800), random_state=1)
fig5 = px.scatter(sample, x="Sales", y="Profit", color="Category", size="Sales",
                  hover_data=["Discount", "Region"], title="Sales vs Profit (sample of orders)")
 
# Chart 6: Segment x Category sales (stacked bar)
seg = fdf.groupby(["Segment", "Category"], as_index=False)["Sales"].sum()
fig6 = px.bar(seg, x="Segment", y="Sales", color="Category", barmode="stack", title="Sales by Segment & Category")
 
row1 = st.columns(3)
row1[0].plotly_chart(fig1, use_container_width=True)
row1[1].plotly_chart(fig2, use_container_width=True)
row1[2].plotly_chart(fig3, use_container_width=True)
 
row2 = st.columns(3)
row2[0].plotly_chart(fig4, use_container_width=True)
row2[1].plotly_chart(fig5, use_container_width=True)
row2[2].plotly_chart(fig6, use_container_width=True)
 
# ---------------------------------------------------------------
# 5. DATA TABLE + DOWNLOAD
# ---------------------------------------------------------------
with st.expander("🔍 View filtered data"):
    st.dataframe(fdf, use_container_width=True)
    st.download_button("⬇️ Download CSV", fdf.to_csv(index=False).encode("utf-8"),
                       "filtered_data.csv", "text/csv")
 
```
